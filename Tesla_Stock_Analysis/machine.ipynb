```python
from sklearn.preprocessing import StandardScaler

X = [
    [20, 30000],
    [25, 5000],
    [30, 40000],
]

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print(X_scaled)
```

```python
from sklearn.preprocessing import MinMaxScaler

X = [
    [20, 30000],
    [25, 5000],
    [30, 40000],
]

scaler = MinMaxScaler()

X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

Encoding - numerical format
1.label encoding -male =0, femal -1 -
2.Ordinal encoding : levels of data low medium high
3. one hot encoder -matrix format convert dta

```python
from sklearn.preprocessing import LabelEncoder

gender = ['Male', 'Female', 'Female', 'Male', 'Female']
encoder = LabelEncoder()
gender_encoded = encoder.fit_transform(gender)
print(gender_encoded)
```

```python
print(encoder.classes_)
```

```python
odinal encoding is a technique used to convert categorical variable into numerical valuess.
```
