```python
import matplotlib.pyplot as plt
```

```python
X = [0, 2, 5, 9, 6]
Y = [2, 8, 9, 6, 3]

plt.plot(X, Y)
plt.xlabel("X-axis label")
plt.ylabel("Y-axis label")
plt.title("Plot Title")
plt.show()
```

```python
X = [0, 2, 5, 9, 6]
Y = [2, 8, 9, 6, 3]

plt.plot(X, Y)
plt.xlabel("X-axis label")
plt.ylabel("Y-axis label")
plt.title("Plot Title")
plt.grid(True, linestyle="--", alpha=1)
plt.show()
```
