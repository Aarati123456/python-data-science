```python
## bar chart
```


-display categorical data using rectangular bars whose length are proportional to the values they represent plt.bar()

```python
students = ["Ram", "Balen", "pabi", "hari"]

marks = [70,85,96,90]
plt.bar(
    students,
    marks,
    width=0.6,
    color="skyblue",
    edgecolor="black",
    linewidth=1.5,
    alpha=0.8
)
plt.xlabel("students")
plt.ylabel("marks")
plt.title("students Marks")
plt.grid(axis="y", linestyle="--",alpha=0.5)
plt.show()

```

```python
students = ["Ram", "Balen", "pabi"]
subjects = ["python", "Java", "R"]
marks = [78, 85, 96]

plt.bar(
    students,
    marks,
    width=0.6,
    color="skyblue",
    edgecolor="black",
    linewidth=1.5,
    alpha=0.8,
    label="marks"
)

plt.xlabel("students")
plt.ylabel("subjects")
plt.title("Students Marks")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.ylim(0, 100)

for i, value in enumerate(marks):
    plt.text(i, value + 1, str(value), ha="center")

plt.legend()
plt.show()
```

```python
Histogram:
Statistical diagram -shows the distribution of numberical data.
-bins
it tells us how many data values fall inside different numerical ranges(interval/bins).
HISTOGRAM = numerical data + bins + frequency
```

```python
marks = [70,80,89,74,75,55,66, 55,89,57,65,96,89,68,96,48,42,43,44,54,65,68,69]
plt.hist(
    marks,
    bins=5,
    color="skyblue",
    edgecolor="black",
    alpha=0.8
)
plt.xlabel("marks")
plt.ylabel("Frequency")
plt.title("distribution of student")
plt.show()
```

```python
4. Scatter plot :
it is used to see the relationship between two numerical variables.
upward trend - positive relationship
downward trend - negative relationship
randomly scaterred - weak / no relations    
    
```

```python
hours = [1, 2, 3, 4, 5, 6, 7]
marks = [45, 50, 55, 62, 68, 75, 82]

plt.scatter(
    hours,
    marks,
    s=100,
    color="skyblue",
    edgecolors="black",
    alpha=0.8,
    marker="o"
)

plt.xlabel("hours studied")
plt.ylabel("marks")
plt.title("Study Hours vs Exam Marks")

plt.show()
```

## Pie-chart
represents how total is divided into different categories or proportions.

```python
categories = ["Food", "Travel", "Study", "Entertaiment"]
expenses = [4000, 2000, 3000, 1000]
plt.pie(
    expenses,
    labels = categories,
    autopct="%1.01f%%",
    startangle=90,
    shadow=False
)
plt.title("monthly expenses")
plt.show()
```

Co-relation :
-it is the statistical measure that tells us how two variables relationship and they move in relation to each other
-coorleation analysis
-it calculates the coorelation using pearson cooeleation coefficient
scenario:
r = -1 , indirect a perfect negative corealation
r=0, indicate a no liner corelation
r = 1, perfect positive corelation

range of value: -1 to +1
r = 0.85 - strong liner association

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Hours_Studied": [1,2,3,4,5,6,7,8],
    "Marks": [35,40,45,50,55,60,70,80],
    "Sleep_Hours": [8,7,7,6,6,5,5,4]
}

df = pd.DataFrame(data)

corr = df.corr()

print(corr)
```

```python
plt.figure(figsize=(7,5))
plt.imshow(corr, cmap="coolwarm", vmin=-1,vmax=1) 
plt.colorbar(label="correlation")
plt.xticks(range(len(corr.columns)),corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)),corr.columns)

for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        plt.text(j,i,f"{corr.iloc[i,j]:.2f}", ha="center")
        
plt.title("corelation")
plt.tight_layout()
plt.show()
```
