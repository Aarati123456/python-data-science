```python
## bar chart
```


-display categorical data using rectangular bars whose length are proportional to the values they represent plt.bar()

```python
students = ["Ram", "Balen", "pabi", "hari"]

marks = [70,85,96,90]
plt.bar(
    students,
    marks,
    width=0.6,
    color="skyblue",
    edgecolor="black",
    linewidth=1.5,
    alpha=0.8
)
plt.xlabel("students")
plt.ylabel("marks")
plt.title("students Marks")
plt.grid(axis="y", linestyle="--",alpha=0.5)
plt.show()

```

```python
students = ["Ram", "Balen", "pabi"]
subjects = ["python", "Java", "R"]
marks = [78, 85, 96]

plt.bar(
    students,
    marks,
    width=0.6,
    color="skyblue",
    edgecolor="black",
    linewidth=1.5,
    alpha=0.8,
    label="marks"
)

plt.xlabel("students")
plt.ylabel("subjects")
plt.title("Students Marks")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.ylim(0, 100)

for i, value in enumerate(marks):
    plt.text(i, value + 1, str(value), ha="center")

plt.legend()
plt.show()
```
