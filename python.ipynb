```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def check_pass_fail(self):
        if self.score >= 40:
            return True
        else:
            return False


student1 = Student('Harry', 85)
did_pass = student1.check_pass_fail()

print(f"Did {student1.name} pass?", did_pass)
```

```python
class Student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks
        self.total = 0
        self.average = 0
        self.grade =" "
    
    def calculate_total_average(self):
        self.total = sum(self.marks)
        self.average = self.total / len(self.marks)
        
    def assign_grade(self):
        if self.average >=90:
            self.grade = "A"
        elif self.average >=80:
            self.grade ="B"
        elif self.average >=70:
            self.grade ="C"
        elif self.average >= 60:
            self.grade = "D"
        else:
            self.grade ="F"
            
            
    def show_result(self):
        print(f"Student name is {self.name}")
        print(f"Student marks is {self.marks}")
        print(f"Student total marks is {self.total}")
        print(f"Student Average marks is {self.average:.2f}")
        print(f"Grade is :{self.grade}")
        
        
name = input("enter the student name")
marks=[]
for i in range(1,4):
    mark=float(input("enter marks of subject"))
    marks.append(mark)
    
student = Student(name,marks)
student.calculate_total_average()
student.assign_grade()
student.show_result()
```

```python
# create a class Circle

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        area = 3.14 * self.radius * self.radius
        return area


radius = float(input("Enter the radius of circle: "))

circle1 = Circle(radius)

area = circle1.calculate_area()

print(f"Area of the circle is: {area:.2f}")
```

```python
#Create a parent class MLmodel with model name and accuracy and a method checkperformace that prints whether the model has a good performance.

class MLmodel:

    def __init__(self, model_name, accuracy):
        self.model_name = model_name
        self.accuracy = accuracy

    def checkperformance(self):
        if self.accuracy >= 80:
            print("Model has good performance")
        else:
            print("Model needs improvement")


model1 = MLmodel("KNN", 85)

print("Model name:", model1.model_name)
print("Accuracy:", model1.accuracy)

model1.checkperformance()
```

```python
#Create a parent class MLmodel with model name and accuracy and a method checkperformace that prints whether the model has a good performance.
#Then again create a child class knn with additional attribute k
#create a method show_details in knn that display the model name accuracy and value of k

class MLmodel:

    def __init__(self, model_name, accuracy):
        self.model_name = model_name
        self.accuracy = accuracy

    def checkperformance(self):
        if self.accuracy >= 80:
            print("Model has good performance")
        else:
            print("Model needs improvement")


class KNN(MLmodel):

    def __init__(self, model_name, accuracy, k):
        super().__init__(model_name, accuracy)
        self.k = k

    def show_details(self):
        print("Model name:", self.model_name)
        print("Accuracy:", self.accuracy)
        print("Value of k:", self.k)


model1 = KNN("KNN", 85, 5)

model1.checkperformance()
model1.show_details()
```
