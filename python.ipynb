Slicing of a list:
: -slicing operator
syntax: variable = list_name[start : end ]


```python
numbers = [10,20,30,40,50]
variable = numbers[2:5]
print(variable)
```

Python list index will omit the end index.
here,2:5 i.e-it prints the item from the 2 nd index up to 4th index. it will never ever return the stop index.

```python
numbers = [10,20,30,40,50]
variable = numbers[:5]
print(variable)
```

extend()-it add all the items of one list to the end of another list

```python
animals = ['dog','cat']
numbers = [1,2]
animals.extend(numbers)
print(animals)
```
